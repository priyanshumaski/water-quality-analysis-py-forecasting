```python
import joblib
from prophet import Prophet

def run_prophet_forecast(df, feature, extra_col, forecast_type='monthly'):
    """
    Run the forecast for a given feature (monthly or yearly).
    The model uses mean, standard deviation, and median for more robust forecasting.
    """
    
    # Step 1: Prepare date column
    df['Date'] = pd.to_datetime(df['Year'].astype(str) + '-' + df['Month'].astype(str) + '-01')

    # Step 2: Group by month to get mean, std, and median per month
    monthly_df = df.groupby('Date').agg({feature: ['mean', 'std', 'median'], extra_col: 'mean'}).reset_index()
    
    # Flatten the multi-level column names
    monthly_df.columns = ['Date', f'{feature}_mean', f'{feature}_std', f'{feature}_median', extra_col.lower()]
    
    # Rename columns for Prophet model
    monthly_df = monthly_df.rename(columns={'Date': 'ds', f'{feature}_mean': 'y', extra_col.lower(): extra_col.lower()})

    # Step 3: Remove missing values if any
    monthly_df = monthly_df.dropna()

    # Step 4: Initialize Prophet model
    model = Prophet(yearly_seasonality=True)
    model.add_seasonality(name='monthly', period=30.5, fourier_order=5)
    model.add_regressor(extra_col.lower())  # Add extra feature (Potability)

    # Step 5: Fit the model
    model.fit(monthly_df)
    joblib.dump(model, 'C:/Users/acer/Downloads/model_waterfbpropehet.pkl')

    # Step 6: Create future dataframe for forecasting
    if forecast_type == 'monthly':
        future_periods = 12  # Forecast next 12 months
    elif forecast_type == 'yearly':
        future_periods = 36  # Forecast next 3 years
    else:
        raise ValueError("forecast_type must be 'monthly' or 'yearly'")

    # Create future dates
    future = model.make_future_dataframe(periods=future_periods, freq='MS')

    # Merge Potability values (we use last known average value)
    last_value = monthly_df[extra_col.lower()].iloc[-1]
    future[extra_col.lower()] = last_value  # Use constant Potability value

    # Step 7: Predict the forecast
    forecast = model.predict(future)

    # Step 8: Plot the forecast
    title_map = {
        'monthly': f"Monthly Forecast of {feature} using {extra_col}",
        'yearly': f"Yearly Forecast of {feature} using {extra_col}"
    }

    # For monthly forecast: Plot only the next 12 months with month-wise labels
    if forecast_type == 'monthly':
        forecast_12 = forecast.tail(12)  # Get only the last 12 months of forecast
        forecast_12['Month'] = forecast_12['ds'].dt.strftime('%b')  # Extract month names

        plt.figure(figsize=(10, 5))
        plt.plot(forecast_12['Month'], forecast_12['yhat'], marker='o', color='teal', label='Forecast')

        # Add value labels above the forecast points
        for i, value in enumerate(forecast_12['yhat']):
            plt.text(i, value, f'{value:.2f}', ha='center', va='bottom', fontsize=9)

        plt.title(title_map['monthly'])
        plt.xlabel("Month")
        plt.ylabel(feature)
        plt.grid(True)
        plt.tight_layout()
        plt.show()

    # For yearly forecast: Plot the full yearly forecast
    elif forecast_type == 'yearly':
        model.plot(forecast)
        plt.title(title_map['yearly'])
        plt.xlabel("Date")
        plt.ylabel(feature)
        plt.grid(True)
        plt.show()


# Features to forecast 
forecast_features = ['ph', 'Hardness', 'Solids', 'Chloramines',
                     'Conductivity', 'Organic_carbon', 'Trihalomethanes', 'Turbidity']

# Run forecast for each feature for both monthly and yearly forecasts
for feature in forecast_features:
    run_prophet_forecast(df, feature, 'Potability', forecast_type='monthly')
    run_prophet_forecast(df, feature, 'Potability', forecast_type='yearly')
```
